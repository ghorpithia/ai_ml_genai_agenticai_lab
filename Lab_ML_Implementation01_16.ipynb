### IQR method

```python
dfiris["SepalWidthCm"]
```

```python
q1 =dfiris["SepalWidthCm"].quantile(0.25)
```

```python
q3 =dfiris["SepalWidthCm"].quantile(0.75)
```

```python
q1
```

```python
q3
```

```python
dfiris["SepalWidthCm"].describe()
```

```python
iqr = q3 - q1
```

```python
iqr
```

```python
lower = q1 - (1.5*iqr)
```

```python
lower
```

```python
upper = q3 + (1.5*iqr)
```

```python
upper
```

```python
( dfiris["SepalWidthCm"] < lower) | (dfiris["SepalWidthCm"] > upper)
```

```python
dfiris[ ( dfiris["SepalWidthCm"] < lower) | (dfiris["SepalWidthCm"] > upper) ]
```
